Notebook minh họa loss cho hồi quy, phân loại nhị phân và đa lớp, cùng một custom loss MAPE có thể tính gradient.

```python
import torch
from torch import nn

# Với cùng một dự đoán, MAE ít nhạy với sai số lớn hơn MSE.
prediction = torch.tensor([0.0, 3.0], dtype=torch.float32)
target = torch.tensor([1.0, 0.0], dtype=torch.float32)
mae = nn.L1Loss()(prediction, target)
mse = nn.MSELoss()(prediction, target)
huber = nn.HuberLoss()(prediction, target)
print(f"MAE={mae.item():.3f} | MSE={mse.item():.3f} | Huber={huber.item():.3f}")
```

```python
# Phân loại nhiều lớp: CrossEntropy nhận logits; LogSoftmax + NLL cho kết quả tương đương.
logits = torch.tensor([[2.0, 0.5, -1.0], [0.2, 1.3, -0.4]], dtype=torch.float32)
class_target = torch.tensor([0, 1], dtype=torch.long)
ce_loss = nn.CrossEntropyLoss()(logits, class_target)
nll_loss = nn.NLLLoss()(nn.LogSoftmax(dim=1)(logits), class_target)
print(f"CrossEntropy={ce_loss.item():.4f} | LogSoftmax + NLL={nll_loss.item():.4f}")
torch.testing.assert_close(ce_loss, nll_loss)
```

```python
# Phân loại nhị phân: BCELoss nhận xác suất sau sigmoid.
probabilities = torch.tensor([[0.8], [0.25]], dtype=torch.float32)
binary_target = torch.tensor([[1.0], [0.0]], dtype=torch.float32)
print(f"Binary cross entropy={nn.BCELoss()(probabilities, binary_target).item():.4f}")
```

```python
# Custom MAPE: target dương để phép chia phần trăm có nghĩa.
def mape_loss(output, actual):
    return torch.mean(torch.abs((actual - output) / actual))

regression_output = torch.tensor([[2.2], [4.5], [7.0]], dtype=torch.float32, requires_grad=True)
regression_target = torch.tensor([[2.0], [5.0], [8.0]], dtype=torch.float32)
custom_loss = mape_loss(regression_output, regression_target)
custom_loss.backward()
print(f"Custom MAPE={custom_loss.item():.4f} | gradient exists={regression_output.grad is not None}")
```

Notebook minh họa các hàm mất mát cho hồi quy và phân loại, rồi viết custom MAPE bằng các phép toán tensor PyTorch. Custom loss trả về một tensor vô hướng để có thể lan truyền gradient.
